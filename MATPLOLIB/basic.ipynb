```python
n = [ 1,2,3,4]
label = ['apple' , 'mango' , 'grapes', 'lichi']
plt.pie(n , labels = label)
plt.show()
```

```python
n = [ 1,2,3,4]
label = ['apple' , 'mango' , 'grapes', 'lichi']
plt.pie(n , labels = label,explode=[0,0,0.2,0],shadow=True)
plt.show()
```

```python
n = [ 1,2,3,4]
label = ['apple' , 'mango' , 'grapes', 'lichi']

plt.pie(n , labels = label,explode=[0,0,0.2,0],shadow=True,autopct='%1.1f%%')
plt.show()
```

```python
colours = np.array(['red', 'yellow', 'purple','green'])
n = [ 1,2,3,4]
label = ['apple' , 'mango' , 'grapes', 'lichi']

plt.pie(n , labels = label,explode=[0,0,0.2,0],shadow=True,autopct='%1.1f%%',colors=colour)
plt.show()
```

```python
colours = np.array(['red', 'yellow', 'purple','green'])
n = [ 1,2,3,4]
label = ['apple' , 'mango' , 'grapes', 'lichi']

plt.pie(n , labels = label,explode=[0,0,0.2,0],shadow=True,autopct='%1.1f%%',colors=colour)
plt.legend()
plt.title("FRUIT DESTRIBUTION")
plt.show()
```
