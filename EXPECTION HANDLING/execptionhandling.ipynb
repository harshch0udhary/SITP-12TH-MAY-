Exception handling is a mechanism in programming used to handle runtime errors gracefully without crashing the program.


| Keyword   | Purpose                       |
| --------- | ----------------------------- |
| `try`     | Code that may cause exception |
| `except`  | Handles exception             |
| `finally` | Executes always               |
| `raise`   | Manually throw exception      |


```python
try:
    pass
except:
    pass
else:
    pass
finally:
    pass
```

```python
try:
    a=int(input("Enter the first number :"))
    b=int(input("Enter the second number:"))
    print(a/b)

except ZeroDivisionError as e:
    print(e)
except ValueError as e:
    print(e)
```

```python
#ELSE BLOCK

try:
    num=int(input("enter your number:"))
except ValueError as e:
    print("INVALID NUMBER")
else:
    print("your input",num)    
```

```python
#FINALLY BLOCK

try:
    num=int(input("enter your number:"))
except ValueError as e:
    print("INVALID NUMBER")
else:
    print("your input",num)   
finally:
    print("mein to hamesa cahluga") 
```

```python
class User:
    def __init__(self, password):
        if len(password) < 8:
            raise ValueError("Password must be at least 8 characters long")
        self.password = password

try:
    user = User("abc")
except ValueError as e:
    print("Error:", e)
```

```python
class User:
    def __init__(self, email):
        if email :
            raise ValueError("EMAIL must be at correct")
        self.email=email

try:
    user = User("abc")
except ValueError as e:
    print("Error:", e)
```

```python
class User:
    def __init__(self, password):
        if '@' not in password:
            raise ValueError("Password must contain '@'")
        self.password = password
        print("Password accepted")

try:
    user = User("abc123")
except ValueError as e:
    print("Error:", e)
```

```python
class User:
    def __init__(self, password):
        if not any(char.isupper() for char in password):
            raise ValueError("Password must contain an uppercase letter")

        if not any(char.islower() for char in password):
            raise ValueError("Password must contain a lowercase letter")

        if not any(char.isdigit() for char in password):
            raise ValueError("Password must contain a digit")

        if not any(char in "@#$%&*" for char in password):
            raise ValueError("Password must contain a special character")

        self.password = password
        print("Password accepted")

try:
    user = User("shjsdkjds15366")
except ValueError as e:
    print("Error:", e)
```
