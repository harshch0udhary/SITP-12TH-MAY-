```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False  
    return True  
num = int(input("enter the number check it is prime or not:"))
if is_prime(num):
    print(f"{num} is a prime number")
else:
    print(f"{num} is not a prime number")
marks= [35,65,75]
marks.append(num)
print("list of marks<<<<<<<:", marks)
```
