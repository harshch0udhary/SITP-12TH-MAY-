```python
#multiple inhertiance by using construtor 

class A:
    def __init__(self,name):
        self.name= name
class B(A):
    def __init__(self,roll_no):
        self.roll_no= roll_no
class C(B):
    def __init__(self,branch):
        self.branch= branch
clg=C(branch="AI")
clg=A(name="HARSH")
clg.name
```
