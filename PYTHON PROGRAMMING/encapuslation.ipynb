```python
#Public Access Specifier in Python

In Python, public members are variables and methods that can be accessed from anywhere:

Inside the class
Outside the class
In derived classes
```

```python
class Collage :
    Collage_name="bkbiet"#public variable
    Collage_garden="yes"#public variable
    Collage_account="34738628238"#public variable

    clg=Collage()
    clg.Collage_name
```

```python
#Private Access Specifier in Python

In Python, private members are variables or methods that are intended to be used only inside the class.

Private members are created using double underscore __ before the variable or method name
```

```python
class Collage :
    Collage_name="bkbiet"     #public variable
    __Collage_garden="yes"    #using __ for private>>>>>>>>
    Collage_account="34738628238"#public variable
clg=Collage()
clg.Collage_account
clg.__Collage_garden
```

```python
Protected Access Specifier in Python

In Python, protected members are variables or methods that are intended to be accessed:

Inside the class
Inside derived (child) classes

Protected members are created using a single underscore _ before the name.
```

```python
class Collage :
    Collage_name="bkbiet"     #public variable
    _Collage_garden="yes"    #using __ for private>>>>>>>>
    Collage_account="34738628238"#public variable
clg=Collage()
clg.Collage_account
clg._Collage_garden
```
