```python
import cv2
img_arr_list = []
img_key = []

for key, values in dct.items():
    for path in values:
        img_arr = cv2.imread(path)
        resized_image_array = cv2.resize(img_arr, (224, 244))
        img_arr_list.append(resized_image_array)
        img_key.append(key)
img_arr_list
img_key
img_dict = {"cats":0,"dogs":1}
encoded_keys = [img_dict[item] for item in img_key]
encoded_keys = np.array(encoded_keys)
encoded_keys
```

```python
X=np.array(img_arr_list)
Y=np.array(encoded_keys)
```

```python
X=X/225.0
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42,
    stratify=Y

)
```

```python
print(X_train.shape)
print(X_test.shape)

print(Y_train.shape)
print(Y_test.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import MaxPooling2D,Flatten,Dense
```

```python
from tensorflow.keras.layers import Conv2D
```

```python
model=Sequential(
    [
        Conv2D(32,(3,3),activation='relu',input_shape=(224,244,3)),
        MaxPooling2D(2,2),
        Conv2D(64,(3,3),activation='relu'),
        MaxPooling2D(2,2),
        Conv2D(64,(3,3),activation='relu'),
        MaxPooling2D(2,2),
        Flatten(),
        Dense(units=64,activation='relu'),
        Dense(units=1,activation='sigmoid')
    ]
)
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
history = model.fit(
    X_train,
    Y_train,
    epochs=10,
    validation_data=(X_test, Y_test),
    batch_size=32
)
```

```python
pred = model.predict(X_test)
```
