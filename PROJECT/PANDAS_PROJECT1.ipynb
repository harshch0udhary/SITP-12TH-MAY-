```python
# Check Duplicates
df.duplicated()
df.duplicated().sum()
```

```python
# Remove Duplicates
df.drop_duplicates(inplace=True)
```

```python
# Check Missing Values
df.isnull().sum()
```

```python
# Remove Missing Values
df.dropna(inplace=True)
```

```python
# Numerical and Categorical Columns
object_columns = df.select_dtypes(include=['object']).columns
print(object_columns)
```

```python
numeric_columns = df.select_dtypes(include=['int64','float64']).columns
print(numeric_columns)
```

```python
# ----------------------
# Visualization Section
# ----------------------

# Satisfaction Count
sns.countplot(x='satisfaction', data=df)
plt.title("Customer Satisfaction Distribution")
plt.show()
```

```python
# Gender vs Satisfaction
sns.countplot(x='Gender', hue='satisfaction', data=df)
plt.title("Satisfaction by Gender")
plt.show()
```

```python
# Age Distribution
sns.histplot(df['Age'], bins=20, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
# Flight Distance Distribution
plt.hist(df['Flight Distance'])
plt.title("Flight Distance")
plt.show()
```

```python
# Line Plot
sns.lineplot(df['Age'])
plt.show()
```

```python
# Scatter Plot
sns.scatterplot(x='Age', y='Flight Distance', data=df)
plt.show()
```

```python
# Box Plot
sns.boxplot(x='satisfaction', y='Flight Distance', data=df)
plt.title("Flight Distance vs Satisfaction")
plt.show()
```

```python
# Inflight Entertainment vs Satisfaction
sns.boxplot(x='satisfaction', y='Inflight entertainment', data=df)
plt.title("Inflight Entertainment vs Satisfaction")
plt.show()
```

```python
# Cleanliness vs Satisfaction
sns.boxplot(x='satisfaction', y='Cleanliness', data=df)
plt.title("Cleanliness vs Satisfaction")
plt.show()
```

```python
# Correlation Heatmap
numeric_df = df.select_dtypes(include=['int64','float64'])

plt.figure(figsize=(10,8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
