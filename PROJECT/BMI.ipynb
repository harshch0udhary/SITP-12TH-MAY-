```python
# BMI CALCULATOR

print("<<<<<<<<<<<<<< WELCOME TO BMI CALCULATOR >>>>>>>>>>>>>>")

class BMI:
    def __init__(self, weight, height):
        self.weight = weight
        self.height = height

    def calculate_bmi(self):
        bmi = self.weight / (self.height ** 2)
        print("YOUR BMI IS :", round(bmi, 2))

        if bmi < 18.5:      #using for underweight
            print("CATEGORY : UNDER WEIGHT")
        elif bmi < 25:      #using for normal weight 
            print("CATEGORY : NORMAL WEIGHT")
        elif bmi < 30:      #using for over weight 
            print("CATEGORY : OVER WEIGHT")
        else:               #using for obese
            print("CATEGORY : OBESE")


name = input("ENTER YOUR NAME : ")  #USER NAME 

if name != "":
    print("WELCOME", name)

    weight = float(input("ENTER YOUR WEIGHT (kg) : "))   #This using for weight 
    height = float(input("ENTER YOUR HEIGHT (m) : "))    #This using for height 

    bmi_obj = BMI(weight, height)

    bmi_obj.calculate_bmi()

    print("<<<<<<<<<<<<<< THANK YOU FOR USING BMI CALCULATOR >>>>>>>>>>>>>>")

else:
    print("INVALID NAME")
```
