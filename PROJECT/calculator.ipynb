```python
a=int(input("ENTER THE FIRST NUMBER:"))
b=int(input("ENTER THE SECOND NUMBER:"))
add=a+b
sub=a-b
mul=a*b
div=a/b
print("ADDING OF Two numbers:",add)
print("SUBTRACTION OF Two numbers:",sub)
print("MULTIPICATION OF Two numbers:",mul)
print("DIVISON OF TWO NUMBER:",div)
if div>=1:
    print("positive divison")
else:
    print("negative subtraction")
```
